## Intelligent systems

An intelligent system can:

- receive information about its environment;
- retain or communicate useful information;
- use that information to adjust its behaviour; and
- improve performance according to a stated objective.

## Collective behaviour and collective intelligence

**Collective behaviour** is any population-level pattern produced by interactions among individuals. A stop–start wave in a dense crowd is one example and need not help the group achieve a shared task.

**Collective intelligence** describes cases in which these interactions help a group process information, make a decision or solve a problem.

This week asks how collective mechanisms can be used for computation and search.

## Optimisation

$$
\min_{x\in\mathcal X} f(x)
$$

- **Candidate** $x$: one possible solution.
- **Solution space** $\mathcal X$: all allowed candidates, including any constraints.
- **Objective** $f(x)$: a numerical measure of candidate quality; here, lower is better.
- **Evaluation budget**: the number of candidates the algorithm may score.

Some optimisation literature uses **fitness** for a score that increases with quality. State the direction whenever that term is used.

The solution space is usually abstract: its coordinates are decision variables rather than physical locations.


## Canonical model 1: ACO

<iframe src="images/aco_network_explorer.html" title="Ant colony optimisation learning a route through a network" style="display:block;width:100%;height:545px;border:0;margin:0 auto;"></iframe>


## Model details

| Component | Definition or choice |
|---|---|
| graph | $G=(V,E)$, with fixed edge costs $d_e>0$, start node $s$ and target node $T$ |
| shared information | pheromone $\tau_e(n)$ on each edge at generation $n$ |
| population | $N$ ants constructing complete routes per generation |
| choice rule | pheromone weight $\alpha$ and fixed-heuristic weight $\beta$, with $\eta_e=1/d_e$ |
| pheromone update | evaporation fraction $\rho$ and deposit scale $Q$ |
| stopping rule | a stated number of generations or evaluations, or a route-cost threshold chosen in advance |



## Choose the next edge

For an ant currently at node $i$, $P(e\mid i)$ is the probability that its next selected edge is $e$:

$$
P(e\mid i)=
\frac{\tau_e^{\alpha}\eta_e^{\beta}}
{\displaystyle\sum_{e'\in\mathcal A(i)}\tau_{e'}^{\alpha}\eta_{e'}^{\beta}}.
$$

$\mathcal A(i)$ is the set of currently allowed outgoing edges. Sampling from this distribution is repeated until the ant reaches the target and has constructed one complete candidate route.


## Evaluate complete paths and update pheromone

If ant $a$ constructs route $R_a$, its total cost is

$$
L_a=\sum_{e\in R_a}d_e.
$$

After all $N$ routes have been evaluated, evaporate pheromone and reinforce the edges used by shorter routes:

$$
\tau_e(n+1)=(1-\rho)\tau_e(n)+
\sum_{a=1}^{N}\Delta\tau_e^{(a)},
\qquad
\Delta\tau_e^{(a)}=
\begin{cases}
Q/L_a,&e\in R_a,\\
0,&\text{otherwise.}
\end{cases}
$$

Lower-cost routes therefore make larger deposits. Pheromone is held fixed while the routes in one generation are constructed.


## Analysis

| Measurement | Interpretation |
|---|---|
| best route cost | solution quality |
| fraction of runs finding a route at or below the chosen cost threshold | reliability across stochastic runs |
| completed routes evaluated before first success | search efficiency |
| pheromone concentration and route diversity | how continued search is distributed across the graph |

## Canonical model 2: PSO

1. Kennedy and Eberhart began with a simplified flock model.
2. A fixed **cornfield vector** marked a target at known screen coordinates.
3. Distance from the target gave each visited position an objective value; particles were not given a direction to the target.
4. Each particle retained its own best position and received the swarm's best recorded position.
5. Position became a candidate solution in an abstract solution space.

As in Week 5, the agents move and exchange social information. Here those processes search an abstract solution space.


<div class="reader-voice slide-quote">
  <div class="reader-voice-quote">very simple concept, and paradigms can be implemented in a few lines of computer code.</div>
  <div class="reader-voice-attr">Kennedy and Eberhart, <a href="https://doi.org/10.1109/ICNN.1995.488968"><em>Particle Swarm Optimization</em></a> (1995)</div>
</div>


## Model details

Each particle $i$ stores four things:

| State | Meaning |
|---|---|
| $x_i$ | current candidate solution |
| $v_i$ | proposed movement through the search space |
| $p_i$ | particle's personal best position |
| $g$ | best position shared by the swarm |

Particles repeatedly move, evaluate the objective and update these records.


## Model specification

| Component | Choice required |
|---|---|
| problem | objective $f(\mathbf{x})$, feasible space and bounds |
| population | particle count $N$ and initial positions and displacements |
| information | global-best or neighbourhood sharing |
| parameters | inertia $w$, personal weight $c_1$, shared weight $c_2$ |
| implementation | boundary rule and any maximum move |
| stopping | evaluation budget or declared objective target |

These choices are part of the algorithm and must be reported with its performance.


## Update displacement and position

For particle $i$ in a $D$-dimensional search space,

$$
x_i^n,\;v_i^n,\;p_i,\;g\in\mathbb R^D.
$$

For every particle and iteration, draw fresh random multipliers $r_{1,i}^n,r_{2,i}^n$ and set

$$
v_i^{n+1}=
\underbrace{w v_i^n}_{\text{inertia}}
+\underbrace{c_1r_{1,i}^n(p_i-x_i^n)}_{\text{personal memory}}
+\underbrace{c_2r_{2,i}^n(g-x_i^n)}_{\text{shared information}},
$$

$$
x_i^{n+1}=x_i^n+v_i^{n+1}.
$$

<p style="font-size:0.72em;color:#5B6780"><strong>Notation:</strong> the coordinate subscript $d$ is suppressed here. The update is applied coordinate by coordinate, with independent $r_{1,i,d}^n,r_{2,i,d}^n\sim\operatorname{Uniform}(0,1)$ draws.</p>


## One PSO iteration

1. Evaluate every current candidate position.
2. Update each particle's personal best.
3. Update the best position shared globally.
4. Draw fresh random multipliers.
5. Update each displacement and position.
6. Apply the stated boundary rule.

Repeat until a satisfactory objective target is reached or the evaluation budget is exhausted. Return the best position found, not the final swarm centroid.


## Exploration and exploitation

Every new objective evaluation can either gather evidence elsewhere or refine what already looks promising.

<div class="two-panel equal-panels compact-panels">
<div class="text-panel"><p><strong>Exploration</strong> means visiting new or weakly sampled regions and preserving differences between candidate solutions.</p></div>
<div class="text-panel"><p><strong>Exploitation</strong> means concentrating evaluations near the best solutions found so far.</p></div>
</div>

Too much exploration wastes evaluations. Too much exploitation produces <strong>premature convergence</strong>: the population agrees before it has searched sufficiently widely.

## ACO and PSO: a comparison

| | ACO | PSO |
|---|---|---|
| candidate | path through a graph | point in a continuous search space |
| memory | pheromone on edges | personal and shared best positions |
| communication | indirect, through the environment | best-so-far record is shared globally |
| main risk | reinforcing a poor route | collapsing too early around a local minimum |

# Framing intelligent systems


An **intelligent system** receives information, uses it to choose between possible actions, and changes its behaviour in ways that help it pursue a goal or respond effectively to changing conditions. This definition includes artificial systems, nervous systems and collectives whose members coordinate without a central controller.


## Two approaches to intelligent systems

<img src="images/intelligent_systems_schism.svg" alt="Complexity science and machine learning as complementary approaches to intelligent systems" style="display:block;width:min(100%,950px);margin:1rem auto">

David Krakauer, President of the Santa Fe Institute, describes a developing scientific **schism**. Machine learning can preserve enormous amounts of detail in a high-dimensional model and make very accurate predictions whose basis is difficult to explain. Complexity science more often searches for a compressed description that exposes a mechanism or useful scale, accepting some loss of predictive power. Here the comparison concerns how each approach studies or builds intelligent behaviour.

| Approach | Main aim | Typical questions about intelligence |
|---|---|---|
| Complexity science | Explain collective and adaptive behaviour using compact models of the important mechanisms and scales. | How does intelligence arise from interactions among many parts? Where is information stored and processed? Which mechanisms allow the system to adapt or make a collective decision? |
| Machine learning and AI | Build systems that predict, classify, decide or act effectively. | What information is needed to predict or classify an outcome? How accurately does the system perform on new cases? Which action best achieves the objective? How can performance improve with experience? |

Artificial intelligence includes several routes to task performance. IBM Deep Blue searched possible chess positions using specialised chess knowledge and a designed evaluation function. Modern neural networks instead fit many parameters from data. Both are judged principally by how well they perform the task.

Collective intelligence sits between the two approaches. We begin with collective mechanisms studied in complexity science and use them to construct search algorithms. The two canonical models show different ways that collective search can store and use evidence. We will ask where information is stored, how it changes later candidate solutions and how well the resulting search performs. Krakauer argues that science needs both prediction and understanding.

## Superorganisms

Interactions among individuals can produce a population-level pattern or response. This is **collective behaviour**. When those interactions allow the group to combine information and use it to make a decision, adapt or solve a problem, the group displays **collective intelligence**.

<div class="reader-voice">
  <div class="reader-voice-quote">In theory at least, individual members of the school can profit from the discoveries and previous experience of all other members of the school during the search for food. This advantage can become decisive, outweighing the disadvantages of competition for food items, whenever the resource is unpredictably distributed in patches.</div>
  <div class="reader-voice-attr">E. O. Wilson, <em>Sociobiology: The New Synthesis</em> (1975)</div>
</div>

Ants, honeybees and termites are often described as **superorganisms**. Their colonies have persistent organisation, division of labour and specialised reproductive roles, and many useful functions belong to the colony as a whole. Each animal still has its own sensory and behavioural capacities. No individual holds all of the colony's information or directs the complete response.

Social-insect colonies combine individual sensing with division of labour and local communication. The reproductive caste produces offspring. Workers coordinate the other colony functions through their behaviour and environmental feedback.

A school of fish also coordinates movement and shares information. It usually lacks the persistent reproductive division of labour and colony-level life cycle that make a social-insect colony function as an integrated biological unit. Individual fish retain the same basic roles and reproduce independently. Collective behaviour alone is therefore insufficient for superorganism status.

At the colony scale, honeybees regulate hive temperature and choose new nest sites. Ants organise traffic, construct nests and living bridges, and assemble into rafts during floods. These are biological examples of **distributed computation**. By distributed computation, I mean computation in which information storage, processing and decision-making are spread across interacting components. The components may be agents, nodes in a network or parts of a physical dynamical system. Collective intelligence and physical reservoir computing are both forms of distributed computation.

### Honeybees communicate by dancing

When a successful honeybee forager returns to the hive, her **waggle dance** can recruit other foragers to the resource. The dance occurs on a vertical comb inside the hive. A waggle run straight upward indicates food in the direction of the Sun. A run angled 30 degrees to the left of vertical indicates food 30 degrees to the left of the Sun's direction. The duration of the waggle run indicates distance. The dance therefore communicates information about a place that is currently outside the hive and out of sight.

Recruited bees visit the advertised source. If it remains profitable, some return and dance for it themselves. Better sources receive more recruitment, while recruitment declines as a source becomes less rewarding. Repeated dances therefore change how the colony distributes its foragers.

### Ants communicate through the environment


Ants coordinate foraging through pheromone traces. A successful forager deposits pheromone while returning to the nest, and later ants detect the trail with their antennae and are more likely to follow it. This produces positive feedback as repeated use strengthens a route. Evaporation weakens information that is no longer reinforced. The balance between these processes determines whether the colony maintains alternative routes or concentrates traffic on one. Information about recent traffic is therefore stored in the environment itself.


Pheromone-based communication is **stigmergy**: agents coordinate indirectly by responding to traces left in a shared environment. The term was introduced by Pierre-Paul Grassé in work on termite construction. A trace can persist after the individual that produced it has moved away, so the environment becomes part of the colony's memory.


## Modelling collective behaviour

The modelling question is which individual observations and responses are sufficient to produce these population-level outcomes. The examples below connect observed ant behaviour to candidate rules and engineered algorithms.

### Shortest paths

Together many ants develop a complex network of pheromone trails, which acts as a shared external memory for the colony. This can allow a colony to favour a shorter route.

In the **double-bridge experiment**, Goss et al. (1989) connected the nest and food source using a bridge that divided into two branches. When the branches had equal length, small chance differences could be amplified until one branch dominated. When one branch was shorter, ants usually concentrated on it because they completed it sooner and reinforced it more rapidly. The result is probabilistic rather than guaranteed. Positive feedback can select a route, while unequal travel times bias that feedback towards the shorter route. See [Self-organized shortcuts in the Argentine ant](https://doi.org/10.1007/BF00462870).

No ant needs a map of the whole network. Route choice emerges from repeated local decisions and pheromone stored in the environment.


### Lane formation

Couzin and Franks studied bidirectional traffic on army-ant trails. Returning ants carrying food and outbound ants respond locally during encounters. These responses can organise three traffic lanes and reduce congestion without assigning individual ants to lanes.


**Data.**

![Couzin lane formation data](images/Couzin_lane_formation_data.png)

Couzin and Franks filmed a section of an army-ant trail and reconstructed individual trajectories. Measurements of speed, acceleration, interaction distance and turning after encounters were used to estimate parameters for the simulation. The panel above is therefore empirical evidence, not model output. It describes both the collective lane pattern and the individual movements from which a candidate rule can be built.

Using data to choose parameter values is **calibration**. It asks which values make the represented ants behave like the observed ants. The model is then evaluated against population-level features that were not directly imposed during calibration: do the calibrated local rules generate directional traffic, distinct lanes and high flow without assigning any ant to a lane? Because calibration and evaluation occur within the same study and system, this is evidence for the proposed mechanism rather than a final, independent validation.

Source: Couzin and Franks (2003), [*Self-organized lane formation and optimized traffic flow in army ants*](https://doi.org/10.1098/rspb.2002.2210).


**Model.**

Each simulated ant has a position, direction and intended direction of travel. It follows a fixed pheromone trail and turns away from nearby ants. Field observations showed that outbound ants turn more sharply during avoidance than returning ants. The model tests how this asymmetry and the strength of directional preference affect traffic flow. Three lanes emerge without assigning any ant to a lane, with returning ants occupying the centre and outbound ants using the margins.

The model is assessed at the collective level: does it reproduce organised three-lane flow and place the measured ant behaviour in a parameter region with low congestion and high traffic flow? # Optimisation

An optimisation problem must specify three things before an algorithm can search it:

- a **candidate solution** $\mathbf x$ and the **solution space** (or search space) $\mathcal X$ containing all allowed candidates.
- an **objective function** $f:\mathcal X\to\mathbb R$ that assigns each candidate a numerical quality score. Some optimisation literature calls this score **fitness**, often with larger values meaning better. Its direction must always be stated.
- any constraints that define the feasible part of $\mathcal X$.

We use the minimisation convention,

$$
\min_{\mathbf x\in\mathcal X} f(\mathbf x).
$$

A quantity that should be maximised can be handled by minimising its negative. In this topic, smaller values of the objective $f$ are better. The objective determines which candidate solutions count as improvements.

The solution space is usually abstract. Its coordinates represent decision variables: controller gains, model parameters, design dimensions or another encoded candidate. A two-dimensional objective **landscape** is a visualisation of the mapping $\mathbf x\mapsto f(\mathbf x)$. The same search space can have dimensions that cannot be drawn.

A local optimum is best only within a nearby region; a global optimum is best across the whole feasible space. Difficult landscapes may contain many local optima, while discrete problems may contain an enormous number of possible routes, schedules or combinations.



We will develop two examples of collective-intelligence optimisation: ant colony optimisation and particle swarm optimisation. These sit within a wider set of population-based and collective-search approaches. In general, we choose the approach that best suits the problem we are tackling.

| Form of one candidate solution | Example problem | Example approaches |
|---|---|---|
| a vector of numerical values | choose model parameters or controller settings | PSO, differential evolution |
| an ordered path, tour or schedule | find a route or order a set of jobs | ACO, genetic algorithms |
| an assignment of agents or resources to tasks | allocate workers, robots or resources | ant/bee allocation, market-based multi-agent methods |
| a set of physical positions or trajectories | cover, monitor or search a region | swarm robotics, foraging algorithms |

Other nature-inspired collective-search algorithms include:

- artificial bee colony optimisation
- grey wolf optimisation
- the firefly algorithm
- the bat algorithm
- cuckoo search
- artificial fish swarm algorithms
- bacterial foraging optimisation
- glowworm swarm optimisation
- the flower pollination algorithm

This is a representative list rather than a complete catalogue. The biological names describe the source metaphor; the important distinction is how each algorithm represents candidate solutions and shares information during the search.


# Canonical model 1: ant colony optimisation

## Model details

**Ant colony optimisation** (ACO) is a family of algorithms inspired by pheromone-mediated route choice. Dorigo and colleagues developed the approach in the early 1990s.

The double-bridge experiment described earlier in the shortest-path section motivates the feedback mechanism used in ACO. Artificial ants construct complete candidate routes. Once the routes have been evaluated, shorter routes add more pheromone to the edges they used. Later ants are then more likely to use those edges, while evaporation allows old information to fade. The same mechanism can be applied to a much larger set of possible routes.

Dorigo, Maniezzo and Colorni's original **Ant System** represented the travelling salesman problem as a weighted graph ([Dorigo, Maniezzo and Colorni, 1996](https://doi.org/10.1109/3477.484436)):

*Given some cities and the distances between every pair of cities, what is the shortest route that visits each city exactly once and returns to the starting city?*

The cities are nodes, the possible journeys between them are edges, and the edge weights are distances. A complete tour is one candidate solution. This is an NP-hard combinatorial optimisation problem.

<div class="optional-reader-flag"><strong>Optional · computational complexity</strong> A <strong>combinatorial optimisation problem</strong> searches for the best choice among a finite set of discrete arrangements, such as routes, orderings or assignments. <strong>NP-hard</strong> means that every problem in NP can be reduced to it in polynomial time. No polynomial-time algorithm is known that solves every instance exactly. This is a worst-case classification. Small or specially structured instances may still be solved exactly, while heuristic methods seek good solutions when exhaustive search is impractical.</div>


For a routing problem, edge costs can encode distance, time, energy, risk or another quantity to minimise. ACO can also represent other discrete or combinatorial problems as sequences of permitted choices. For scheduling, a node may represent a partial schedule and an edge the placement of the next task.


**Initialise the network and population.** Let $G=(V,E)$ be a weighted graph. Each edge $e$ has a fixed cost $c_e$. Define a problem-specific **heuristic** $\eta_e$ that indicates which local choices appear promising without guaranteeing the best complete path. Here $\eta_e=1/c_e$, so locally shorter edges have larger heuristic values. The heuristic remains unchanged during the search.

Each edge also has a pheromone level $\tau_e(n)$, which records information accumulated from earlier routes. Set $\tau_e(0)=\tau_0>0$ on every permitted edge. Using the same small initial value gives no route an initial advantage while allowing every edge to be selected.

Place $N$ ants at the start node and specify the destination. Choose the pheromone weight $\alpha$, heuristic weight $\beta$, evaporation fraction $\rho$ and deposit scale $Q$. State a stopping rule, for example a fixed number of generations or objective evaluations, or finding a route with cost at or below a threshold chosen in advance.


**Construct candidate paths.** At node $i$, ant $a$ chooses among its currently allowed outgoing edges $A_a(i)$. For $e\in A_a(i)$, use

$$
P_a(e\mid i)=\frac{\tau_e^{\alpha}\eta_e^{\beta}}{\displaystyle\sum_{e'\in A_a(i)}\tau_{e'}^{\alpha}\eta_{e'}^{\beta}}.
$$

This multiplicative rule favours edges supported by both accumulated pheromone and problem-defined information. The exponents $\alpha$ and $\beta$ control their relative influence. Normalising the scores gives a probability distribution, so alternatives can still be sampled. Additive scores, rank-based sampling and greedy rules are also possible.

Repeat the choice until the ant reaches the destination.


**Evaluate complete paths and update pheromone.** An edge choice is only one step in a candidate solution. If ant $a$ constructs the complete path $R_a$, define its objective value by

$$
L_a=\sum_{e\in R_a}c_e.
$$

Lower $L_a$ is better. For a scheduling or allocation problem, $L_a$ might instead combine delay, resource use and penalties for violated constraints. The objective must be stated for the application.

Hold $\tau_e(n)$ fixed while all $N$ ants construct complete paths in generation $n$. After every path has been evaluated, update each edge using

$$
\tau_e(n+1)=(1-\rho)\tau_e(n)+\sum_{a=1}^{N}\Delta\tau_e^{(a)},
\qquad
\Delta\tau_e^{(a)}=
\begin{cases}
Q/L_a,&e\text{ lies on path }a,\\
0,&\text{otherwise.}
\end{cases}
$$

The common deposit $Q/L_a$ converts a smaller objective value into stronger reinforcement. The **deposit scale** $Q$ sets the overall strength of reinforcement relative to existing pheromone and evaporation. This scaling is useful because the numerical value of $1/L_a$ depends on how route costs are measured. If $Q$ is too small, learning is slow. If it is too large, recently used routes can dominate almost immediately. Evaporation allows older evidence to fade, preventing permanent dominance and allowing route preferences to change if costs change. This is a synchronous, generation-level update. ACO has no physical time step $\Delta t$ unless a particular application adds one.

## Analysis

ACO returns the lowest-cost complete route found during the search. Further generations allow the population to test alternatives and reuse useful edges while the best route remains uncertain. Convergence onto one route is therefore not sufficient evidence of success. It may indicate useful concentration or early lock-in.

Different measurements answer different questions:

| Measurement | Interpretation |
|---|---|
| best route cost | quality of the best solution found |
| fraction of runs reaching a chosen cost threshold | reliability across stochastic runs |
| completed routes evaluated before first reaching the threshold | search efficiency |
| pheromone concentration and route diversity | where subsequent search is concentrated |

### Interpreting ACO results

The travelling-salesman results use a separate ten-city analysis benchmark and a budget of 100 completed tours. They do not describe the explorable network, nor do they establish parameter values that will work best for other problems. The broad result is consistent with the original [Ant System experiments](https://doi.org/10.1109/3477.484436): combining pheromone reinforcement with strong problem-specific heuristic information produced better tours than either source of information alone. The original paper also found that the quality-weighted Ant-cycle update outperformed deposits that ignored complete-tour quality.

ACO is stochastic and does not guarantee a globally optimal solution. Evidence should therefore include repeated runs, the quality of the best route and the number of completed routes evaluated before success. Pheromone concentration is not sufficient evidence because it can indicate useful concentration or early lock-in. Later methods address this directly: [Ant Colony System](https://doi.org/10.1109/4235.585892) changes when and how pheromone is updated, while [MAX–MIN Ant System](https://doi.org/10.1016/S0167-739X(00)00043-1) bounds pheromone levels and reinforces only selected high-quality tours.

Performance also depends on how the problem is represented and on the information supplied by the heuristic. The algorithm should be compared with independent sampling and with simpler or exact methods where these are available.

# Canonical model 2: particle swarm optimisation

## Model details

Kennedy and Eberhart began with a simplified flock simulation in the same broad artificial-life tradition as Reynolds' Boids. They removed collision avoidance, cohesion and detailed steering. Movement instead responded to promising positions found by the particle and by others.

An early version placed a fixed **cornfield vector** $c=(100,100)$ in a two-dimensional space. Each visited position $x=(x_1,x_2)$ received the objective value

$$
f(x)=\lVert x-c\rVert_2.
$$

The point $c$ uniquely minimises $f$. The designer knew $c$ in order to calculate the objective value, but a particle was not given $c$ or the direction $c-x$. It could only evaluate positions it visited. PSO only requires these objective values, so it can also be used when derivatives of the objective are unavailable.

The **solution space** is the set of possible positions. Each position represents a candidate solution. The **objective landscape** assigns the value $f(x)$ to each candidate. For this example it is a cone-shaped basin centred on $c$, representing candidate quality rather than physical terrain.

<div class="discussion-marker"><img src="images/discussion_marker.svg" alt="Discussion prompt"><span>Vicsek uses social information to generate collective motion, while PSO uses it to guide a search. What changes when shared information is used to find a solution rather than coordinate movement?</span></div>


> *“very simple concept, and paradigms can be implemented in a few lines of computer code.”* — Kennedy and Eberhart, [“Particle Swarm Optimization” (1995)](https://doi.org/10.1109/ICNN.1995.488968)


**Initialise the swarm.** State the objective $f$, feasible bounds, particle count $N$, coefficients, communication topology, boundary rule and stopping condition. Draw each initial position $\mathbf{x}_i^0$ and displacement $\mathbf{v}_i^0$. Set the initial personal record to $\mathbf{p}_i=\mathbf{x}_i^0$, then select the best available personal record as $\mathbf{g}$.


**Evaluate the candidates.** At each visited position, compute $f(\mathbf{x}_i)$. For a minimisation problem, update the personal record when

$$
f(\mathbf{x}_i^n)<f(\mathbf p_i).
$$

In global-best PSO, the shared record is

$$
\mathbf g=\arg\min_{\mathbf p_i} f(\mathbf p_i).
$$

Standard global-best PSO makes $\mathbf g$ available to every particle. Local-best variants restrict sharing to a neighbourhood network.


**Update displacement and position.** Let $n$ denote the iteration number. For particle $i$ and coordinate $d$,

$$
v_{i,d}^{n+1}=
\underbrace{w v_{i,d}^{n}}_{\text{inertia}}
+\underbrace{c_1 r_{1,i,d}^{n}\bigl[p_{i,d}-x_{i,d}^{n}\bigr]}_{\text{personal memory}}
+\underbrace{c_2 r_{2,i,d}^{n}\bigl[g_d-x_{i,d}^{n}\bigr]}_{\text{shared information}},
$$

where $r_{1,i,d}^{n},r_{2,i,d}^{n}\sim\operatorname{Uniform}(0,1)$. Then

$$
x_{i,d}^{n+1}=x_{i,d}^{n}+v_{i,d}^{n+1}.
$$

The conventional ‘velocity’ is a displacement per iteration. The random factors are drawn independently for every particle, coordinate and iteration.


The three terms have distinct roles:

- **inertia:** $w$ controls how strongly the previous displacement persists
- **personal memory:** $c_1$ controls attraction towards the particle's own best position $\mathbf p_i$
- **shared information:** $c_2$ controls attraction towards the best position $\mathbf g$ available through the communication network.

These coefficients do not need to lie in $[0,1]$ or sum to one. Along with the random multipliers, they affect how widely the particles search the objective landscape.


Repeat evaluation and updating until a satisfactory objective target is reached or the declared iteration or evaluation budget is exhausted. Return the best recorded position. Initial conditions and fresh random multipliers make different seeded runs follow different paths, so performance should (of course) be assessed across repeated runs.

## Analysis

The same particle behaviour can look impressive on one objective landscape and fail on another. Evidence therefore depends on an explicitly stated objective, bounds and evaluation budget. PSO seeks a high-quality solution.

Three measurements answer different questions. Let $R$ be the number of runs, $f_{\mathrm{target}}$ the chosen objective threshold and $\mathbf{x}^{(e)}$ the $e$th candidate position evaluated.

| Measurement | Definition | Interpretation |
|---|---|---|
| fraction of runs finding the target | $\displaystyle F=\frac{1}{R}\sum_{r=1}^{R}\mathbb{1}\!\left[\min_{i,n} f(\mathbf{x}_{i,r}^{n})\leq f_{\mathrm{target}}\right]$ | reliability across stochastic runs |
| number of candidate positions evaluated before the target is first reached | $\displaystyle E_{\mathrm{target}}=\min\left\{e:f(\mathbf{x}^{(e)})\leq f_{\mathrm{target}}\right\}$ | search efficiency |
| swarm diversity or concentration | $\displaystyle D(n)=\frac{1}{N}\sum_{i=1}^{N}\lVert\mathbf{x}_i^n-\bar{\mathbf{x}}^n\rVert$ | how widely the remaining search effort is distributed |

If a run does not reach the target, report that $E_{\mathrm{target}}$ exceeded the evaluation budget.

The shared best draws particles towards a promising region, while inertia, personal bests and randomness allow other positions to be tested. The swarm can still settle around a local minimum, so concentration alone does not show that a good solution has been found. Swarm diversity should be interpreted alongside the best objective value.

The same implementation is examined at three levels. The objective is the irregular two-dimensional landscape used in the explorable, the evaluation budget is 100 updates, and the success target is $f(\mathbf{x})<10^{-8}$.

### What does population size change?

With $N=1$, there is no population across which discoveries can spread. In this implementation the lone particle begins with its current position as both personal and shared best, so its initial search is driven mainly by its starting displacement. It is therefore a deliberately weak baseline, not a general model of independent search.

Figure 7.6 fixes the total number of objective evaluations, so larger swarms receive fewer iterations. It shows a population-size trade-off: a modest swarm is more reliable than the one-particle baseline, but further increases do not necessarily help because each particle receives fewer updates. It does not isolate the effect of social information. That requires a matched population in which the sharing rule is changed while the number of starting points and objective evaluations remain fixed.


### Interpreting PSO results

A swarm can concentrate around a local optimum, so convergence alone is not evidence of success. Report objective error, success rate and evaluations to a declared target across repeated runs, and interpret the best objective value alongside swarm diversity.

Established results show the same trade-off: stronger social connectivity spreads discoveries quickly but can reduce diversity, while smaller neighbourhoods often converge more slowly and can be more robust on difficult multimodal problems. The inertia and acceleration coefficients also affect whether trajectories contract, oscillate or become unstable. See [Shi and Eberhart](https://doi.org/10.1109/ICEC.1998.699146), [Clerc and Kennedy](https://doi.org/10.1109/4235.985692), and [Kennedy and Mendes](https://doi.org/10.1109/CEC.2002.1004493).


# ACO and PSO: a comparison

| | Ant colony optimisation | Particle swarm optimisation |
|---|---|---|
| Typical problem | routes, tours and other combinatorial choices | continuous or vector-valued optimisation |
| Search environment | a graph of discrete alternatives | a numerical search space |
| Individual state | the partial path of one ant | position, velocity and personal best |
| Shared information | pheromone stored on edges | the best position reported by the swarm |
| Communication | indirect, through the environment | direct access to a shared record |
| What is shared? | results: successful routes alter pheromone | results: best positions are retained and reported |
| Forgetting | pheromone evaporates | not present unless added explicitly |

ACO and PSO are stochastic population-based methods. Each evaluates several candidate solutions and uses information from the population to guide later search. They do not guarantee a global optimum, so performance must be assessed across repeated runs.

ACO adds **environmental memory** to the modelling toolkit. Information persists outside any individual, changes what later agents encounter, and decays through evaporation. PSO instead stores memory inside particles and shares a best-so-far position. Both are principally forms of **result sharing**: agents search separately, then make discoveries available to others. This differs from **task sharing**, in which a problem is divided into distinct jobs assigned to different agents.

In standard global-best PSO, the best record is aggregated and made available to every particle, so communication is global. A local-best PSO restricts sharing to a neighbourhood graph and is more decentralised. Both families balance exploration and exploitation while representing candidate solutions and exchanging information in different ways. Standard global-best PSO is population-based and centrally aggregates one piece of information. Swarm algorithms can use local or global communication, leaders or shared records, and identical or heterogeneous agents.


# Limits of collective intelligence

Combining individual knowledge into a useful group decision is not straightforward—especially with humans! People may have different objectives, unequal influence, strategic incentives and correlated information. They can also change their behaviour once they understand the rules.

- **Mackay (1841):** [*Memoirs of Extraordinary Popular Delusions and the Madness of Crowds*](https://www.gutenberg.org/ebooks/24518) is an early account of how imitation and shared belief can amplify collective error. It is historical commentary rather than experimental evidence.
- **Lorenz et al. (2011):** even mild social influence reduced the diversity of estimates without improving collective accuracy, while increasing participants’ confidence. The group became more similar and more confident without becoming more accurate. See [*How social influence can undermine the wisdom of crowd effect*](https://doi.org/10.1073/pnas.1008636108).
- **Woolley et al. (2010):** group performance was only weakly related to the average or maximum intelligence of members and was associated with social sensitivity and a more even distribution of conversational turns. See [*Evidence for a collective intelligence factor in the performance of human groups*](https://doi.org/10.1126/science.1193147).

Together, these studies place collective intelligence in the whole feedback structure: the individuals, their objectives, the information they exchange, the environment that stores or filters that information, and the rule used to turn many contributions into one outcome.

Collective decisions can improve or degrade group performance. [Researchers at Unanimous](https://unanimous.ai/swarm/) predicted the exact Kentucky Derby finish of Nyquist, Exaggerator, Gun Runner and Mohaymen, a superfecta that bookmakers had at 540:1.

![Unanimous Kentucky derby](images/Unanimous_Kentucky_derby.gif)

Image: [News.com.au](https://www.news.com.au/technology/innovation/design/swarm-intelligence-predicted-a-540-to-one-long-shot-and-could-be-the-future-of-decision-making/news-story/955f67fd78fe9c73665395637a62840c).

They asked 20 avid racing fans from across the country to pick the race winner. Participants moved a shared “puck” in sync to indicate answers, converging on a group choice that represented their collective conviction.


<div style="background-color: #ffff88; padding: 1em; border-left: 5px solid #fff102;">
For discussion...

Whether we follow the majority or trust experts is shaped by our networks. In general, more bridges allows for more diverse evidence and better collective decisions.

Beyond **Homo Economicus**: People act under imperfect information and cognitive limits (bounded rationality). Decision strategies are context-dependent, not one-size-fits-all. Combining the 
</div>


# Canonical models in pseudocode

The two algorithms share a repeated search cycle while representing candidates and memory differently.

## Ant colony optimisation

```text
INPUT weighted graph G=(V,E) with edge costs d_e
      start node s and target node T
      ant count N and a small initial pheromone tau_0>0
      pheromone weight alpha and heuristic weight beta
      evaporation fraction rho and deposit scale Q
      maximum number of generations K and random seed
DEFINE fixed heuristic eta_e=1/d_e for every allowed edge
INITIALISE pheromone tau_e=tau_0 on every allowed edge
           # tau_0>0 keeps all allowed edges selectable and the normalisation denominator non-zero
FOR generation n=1,...,K
    FOR each ant
        START at the start node with an empty route
        WHILE the target has not been reached
            IDENTIFY all allowed outgoing edges A(i)
            CALCULATE score tau_e^alpha eta_e^beta for every e in A(i)
            NORMALISE these scores across A(i) to obtain probabilities
            SAMPLE one edge from that probability distribution
            APPEND the edge and move to its destination node
        END WHILE
        EVALUATE the complete route cost
    END FOR
    EVAPORATE pheromone on every edge
    DEPOSIT pheromone on used edges according to complete-route quality
    RECORD best route, routes evaluated, route diversity and pheromone
END FOR
RETURN the best complete route found and recorded summaries
```

With $N$ ants and $K$ generations, this version evaluates $NK$ complete routes.

## Particle swarm optimisation

```text
INPUT objective f and feasible bounds X
      particle count N
      inertia w, personal weight c_1 and shared weight c_2
      position and displacement initialisation rules
      boundary rule and maximum displacement if used
      maximum number of iterations K and random seed
INITIALISE particle positions and displacements
EVALUATE every position
SET each personal best to its initial position
SET the global best to the best evaluated position
FOR iteration n=1,...,K
    DRAW personal and shared random coefficients
    UPDATE every velocity from retained motion, personal best and global best
    UPDATE every position and apply the declared boundary rule
    EVALUATE every new position
    UPDATE improved personal bests and the global best
    RECORD best objective, evaluations and swarm diversity
END FOR
RETURN the global best position and recorded summaries
```

The initial evaluation and $K$ updates evaluate $N(K+1)$ candidate positions.


# Scope and connections

Weeks 5–7 have used the same broad framework to ask quite different questions. An agent-based model keeps the components distinct and specifies how each one updates based on its own state, its environment and information obtained from other components.

| Model | Each agent retains | Interaction or shared information | Collective question |
|---|---|---|---|
| Vicsek | position and heading | local alignment with nearby headings, plus noise | when does coherent motion emerge? |
| Kuramoto | phase and natural frequency | phase coupling, usually with the whole population | when can coupling overcome differences between internal clocks? |
| ACO and PSO | a path or candidate solution, plus some form of memory | pheromone in the environment or a shared best position | can a population search for a useful solution? |

This is an incredibly flexible framework. We get to decide what each agent retains, who it interacts with and how it updates. We also decide whether agents differ and how randomness enters. These are claims about how the system works, not just coding decisions.

The same framework has been used to study very different systems:

- [Helbing, Farkas and Vicsek (2000)](https://doi.org/10.1038/35035023) give pedestrians desired velocities, avoidance and contact forces to study crowd escape and pressure.
- [Deffuant et al. (2000)](https://doi.org/10.1142/S0219525900000078) let two agents compromise only when their opinions are sufficiently close, producing clusters of opinion through bounded confidence.
- [Epstein (2002)](https://doi.org/10.1073/pnas.092080199) combines grievance, perceived risk and local policing in an agent-based model of civil violence.
- [O'Keeffe, Hong and Strogatz (2017)](https://doi.org/10.1038/s41467-017-01190-3) couple spatial motion to oscillator phase, joining the moving agents of Week 5 to the internal clocks of Week 6.


## Swarm robotics

Swarm robotics is no longer only a science-fiction idea. Researchers are building and testing groups of drones, miniature robots and modular robots that can explore an area, locate a source, monitor changing conditions, transport objects or assemble structures. A group can cover more ground than one robot, continue when some members fail and adapt its allocation of effort as conditions change.

The collective behaviour still has to be engineered. Robots may coordinate through local sensing, messages or traces left in the environment. Some systems also use a shared controller. The communication architecture determines what information is available and how quickly discoveries spread through the group.

Moving from a numerical optimiser to physical robots also makes the problem harder. Sensing is noisy, messages are delayed or lost, robots collide, batteries run down and hardware fails. Useful systems must therefore be assessed by task performance, robustness, energy use, computation, communication and safety.

The [2024 survey by Wang et al.](https://doi.org/10.1016/j.cja.2024.03.019) reviews models and control methods connecting biological collective motion with robotic swarms. The [Nature collection on swarm intelligence](https://www.nature.com/collections/cgbgjbahac) includes work on miniature robots, drones, modular systems, collective transport, exploration and assembly. Observations of ants, bees and other natural collectives inspire models; the models help us identify and test possible mechanisms, and those mechanisms can then be embedded in engineered systems that operate in the physical world.
