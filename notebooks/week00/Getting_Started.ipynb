### Questions to return to

The canonical model or models change each week, but the questions used to examine them do not. They follow a model from the system being studied to the claim eventually made:

<div class="reader-route">
  <div class="reader-route-label">A recurring model audit</div>
  <div class="reader-route-body">System → representation → implementation → behaviour → evidence → claim</div>
</div>

1. **What are we trying to understand?** What is the question, and what would count as a useful answer?
2. **Where does the system end and the model begin?** Which parts belong to the system being represented, and which are choices made by the modeller?
3. **What is retained, simplified or omitted?** Which details are essential for the question, and which have been deliberately left out?
4. **What produces change?** Identify the state, initialisation, dynamics, interactions, boundaries or network, update order and stopping condition.
5. **Has the model been implemented as intended?** Check limiting cases, known results, invariants, numerical resolution and intermediate states. This is **verification**.
6. **What should we inspect or measure?** What is retained, and what is lost, when detailed behaviour is replaced by a summary?
7. **What is the comparison?** Use a baseline, alternative mechanism, parameter value, system size or null model that makes the result interpretable.
8. **How robust is the result?** Examine seeds, initial conditions, parameters, numerical choices, boundaries and finite system size where they matter.
9. **Does the model help with the original question?** Compare with observations or accepted behaviour where possible, and state the evidence supporting its use. This is **validation**.
10. **What can we reasonably claim?** State limitations, unresolved choices and alternative mechanisms that could produce similar behaviour.

A correct implementation does not guarantee a valid model, and a model that reproduces one observation has not necessarily identified the mechanism that produced it.

Different weeks emphasise different parts of this audit. Do not force every model through all ten questions in equal detail. Return to the full list when planning or reviewing a project. For a quicker check, ask:

- What is represented?
- What has been left out?
- How was the implementation checked?
- What was measured?
- What was it compared with?
- What claim is supported?

## Pseudocode

Pseudocode describes computational logic without tying it to Python syntax. There is no required style for this unit. A useful version makes the order of operations and important decisions visible; a complete version must be precise enough that another person could implement the same model without guessing.

For a complete simulation, state or name the inputs that determine: the world or domain and its size; boundary conditions; stored state; parameter values; random initialisation and seed; neighbourhood or interaction set; update order and clock; stopping rule; and returned output. A named input such as `world_size` is sufficient when its value is intended to be varied, provided its meaning is unambiguous. If the pseudocode describes only one update rather than the full simulation, label it that way.

Before treating pseudocode as a reproducible specification, check six questions: **How is the model initialised? What state determines each update, and how does it evolve? What can interact across a boundary or network? In what order are updates made? When does the run stop? What is recorded, and how often?**

A model specification and pseudocode do related but different work. The specification states what the world contains and what the symbols and parameters mean. The pseudocode states what is done, in what order, and when the process stops. They can be presented together; all essential rules should still be clear.

Recording belongs to the experimental protocol rather than always to the model dynamics. State what is returned and when it is sampled: every state, selected observables at fixed intervals, or only the final state. Intermediate states are needed for animation, transient analysis, some stopping conditions, and later summaries. Otherwise, a final state may be enough. Keeping simulation and measurement separate makes that choice visible.

Each Reader topic ends with clearly labelled pseudocode for the canonical model. It is often repeated in the workshop so that the specification is available beside the implementation. Across the unit, the examples become more technical and use several legitimate forms. The aim is to learn how to communicate a reproducible algorithm, not to memorise one notation. Choose the form that exposes the decisions and dependencies that matter for the problem.
