Game theory was first formalised by John von Neumann and Oskar Morgenstern in their landmark 1944 book *Theory of Games and Economic Behavior*. It gained prominence during the Cold War, when strategists sought mathematical tools to analyse conflict, negotiation, and deterrence.

Game theory is the quantitative study of strategic interaction: players choose among available actions, and each player’s payoff depends on the combination of choices. Dependence on others alone is not enough; Vicsek particles respond to their neighbours but are not choosing actions according to payoffs. Game theory is therefore a natural companion to complex systems in which local decisions by people, organisms, firms or institutions shape collective behaviour.

# From collective behaviour to strategic interaction

Earlier canonical models showed how local interactions produce collective behaviour. They did not necessarily give agents a choice about whether to contribute to that behaviour. Three distinctions help us identify what those models included and what game theory adds.

Coordination means that interaction produces an organised relationship among agents’ behaviours or states—for example, aligned headings, synchronised phases or complementary timing. Cooperation means contributing towards a shared benefit, while negotiation means exchanging proposals or concessions when interests differ. These categories can overlap.

# Real-world motivation


## From shared to conflicting objectives

Game theory studies strategic interaction: players choose among available actions, and their payoffs depend on the combination of choices.

- Coordination: interaction organises agents’ behaviours or states into a collective pattern; local alignment orders flock headings and phase coupling synchronises oscillator rhythms.
- Cooperation: in Week 7, ants and particles share information towards one search objective; contribution is built in.
- Negotiation: exchanging proposals or concessions. None of these models includes bargaining.

Here we ask how individual incentives shape collective outcomes. We begin with rock–paper–scissors and use its familiar rules to introduce strategic choice.


## Start with one rock–paper–scissors game

An **action** is one choice available to a player. A **payoff** (or **utility**) measures how that player values an outcome, with higher values preferred. One occasion on which players choose actions and receive payoffs is one play, or encounter, of a game. Here two players simultaneously choose rock, paper or scissors without seeing the other’s choice. Rock beats scissors, scissors beats paper, and paper beats rock; matching actions draw. Give a win payoff (+1), a loss payoff (-1), and a draw payoff (0):

| Player 1 \ Player 2 | Rock | Paper | Scissors |
|---|---:|---:|---:|
| Rock | $(0,0)$ | $(-1,+1)$ | $(+1,-1)$ |
| Paper | $(+1,-1)$ | $(0,0)$ | $(-1,+1)$ |
| Scissors | $(-1,+1)$ | $(+1,-1)$ | $(0,0)$ |

Read a row as Player 1’s action and a column as Player 2’s action. The ordered pair gives $(u_1,u_2)$: Player 1’s payoff first, then Player 2’s. Rock–paper–scissors is a **zero-sum game**: the two payoffs always add to zero, so one player's gain is the other's loss. It is also a **symmetric game**: exchanging the players’ roles exchanges their payoffs without changing the rules. No single action is best against everything. Each action defeats one alternative and is defeated by another.

# Modelling games

A game is a representation of strategic decisions: the outcome for each player depends on what all players do.

A **rational agent** has consistent preferences over outcomes and chooses an available action that maximises their payoff, given their information and beliefs about the other players. When outcomes are uncertain, we assume the agent maximises **expected payoff**: the average payoff weighted by the probabilities they assign to those outcomes. Payoffs represent what the agent values.

For example, in rock–paper–scissors, paper is rational if the opponent is certain to choose rock. Rationality does not require knowing the opponent’s action in advance: the agent maximises expected payoff using their beliefs. The agent remains motivated by the payoff specified in the model. Representing altruism, fairness or concern for others requires a richer payoff function and is an additional modelling choice.

A one-shot strategic game is specified by

$$
G=(P,\{A_i\}_{i=1}^{n},\{u_i\}_{i=1}^{n})
$$

Here $P=\{1,\ldots,n\}$ is the **player set**, and an **action set** $A_i$ lists player $i$’s available actions. The action sets determine $S=A_1\times\cdots\times A_n$; each member is an **action profile** $\mathbf{a}=(a_1,\ldots,a_n)$. Each **payoff function** $u_i:S\to\mathbb{R}$ assigns player $i$ a utility for every profile.

## Rational agents

A **rational agent** has consistent preferences and chooses an available action that maximises their payoff, given their information and beliefs. Under uncertainty, we assume they maximise **expected payoff**: outcomes’ payoffs weighted by their believed probabilities.

In rock–paper–scissors: if the opponent is certain to choose rock, choose paper.

The agent maximises the payoff specified by the model. **Self-regarding preferences** value only the agent's own outcome; social preferences require a richer payoff function.


Rock–paper–scissors therefore instantiates Equation (1) as $P=\{1,2\}$ and $A_1=A_2=\{\text{rock},\text{paper},\text{scissors}\}$. The payoff table specifies $u_1$ and $u_2$ for every action profile; for example, $u_1(\text{rock},\text{paper})=-1$ and $u_2(\text{rock},\text{paper})=+1$.

Equation (1) specifies incentives. Population structure, learning, movement and repeated interaction are additional modelling choices that determine how those incentives unfold through time.


## Strategies

A **strategy** is a complete rule for how a player acts. In a one-shot game, a **pure strategy** selects one action with certainty. In a repeated game, a strategy can instead prescribe an action for each possible history of earlier play.

## Payoff matrices

A **payoff matrix** displays both players’ payoffs for every action pair. This is the **normal form** of the game: players, available strategies and their payoffs, without a sequence-of-moves diagram.

Here Player 1’s two actions are $\alpha_1,\alpha_2$ and Player 2’s are $\beta_1,\beta_2$. The subscripts distinguish actions, not players. Rows select Player 1’s action, columns select Player 2’s, and each entry lists Player 1’s payoff first:


$$
\begin{array}{c|c|c|}
 & \text{Player 2: } \beta_1 & \text{Player 2: } \beta_2 \\
\hline
\text{Player 1: } \alpha_1 & (u_1(\alpha_1,\beta_1), u_2(\alpha_1,\beta_1)) 
                           & (u_1(\alpha_1,\beta_2), u_2(\alpha_1,\beta_2)) \\
\hline
\text{Player 1: } \alpha_2 & (u_1(\alpha_2,\beta_1), u_2(\alpha_2,\beta_1)) 
                           & (u_1(\alpha_2,\beta_2), u_2(\alpha_2,\beta_2)) \\
\hline
\end{array}
$$

- Player 1’s actions label the rows.
- Player 2’s actions label the columns.
- Each cell of the matrix corresponds to one strategy profile $s\in S$, and the entries show the utilities $(u_1,u_2)$ corresponding to that action pair.

This compact form is called the normal form of a game.

# Model details


## Prisoner's dilemma

*Two suspects are arrested after a burglary. The police can prove only the lesser offence of trespassing unless one suspect testifies against the other. They are interrogated separately and cannot communicate.*

Their sentences depend on both choices.

## Sentence outcomes

| What happens | Sentence |
|---|---|
| One testifies; the other stays silent | The testifying suspect goes free; the silent suspect serves 3 years |
| Both testify | 2 years each |
| Both stay silent | 1 year each |

Each suspect can shorten their own sentence by testifying, whatever the other does, although both are better off if both stay silent. We now represent this situation as a game.

## Setting it up as a game

This is a **one-shot game**: each prisoner makes one choice, with no later rounds.

| Component | Specification |
|---|---|
| Players | $P=\{1,2\}$ |
| Actions | $A_1=A_2=\{C,D\}$, where $C$ means cooperate by staying silent and $D$ means defect by testifying |
| Payoffs (utilities) | $u_1$ and $u_2$ are negative prison sentences, so a less negative value is better |

Because there is only one decision, choosing $C$ or $D$ is also a **pure strategy**.

> Notation warning: here $C$ means cooperate (stay silent) and $D$ means defect (testify). Some versions instead use $C$ for confess and $D$ for don't confess—the letters then mean the exact opposite. Always check the action definitions.

### Action profiles

The set of **action profiles** contains all possible action pairs:

$$
S=\{C,D\}\times\{C,D\}=\{(C,C),(C,D),(D,C),(D,D)\}
$$

For each profile in $S$, the payoff functions $u_1$ and $u_2$ give the two prisoners' utilities.

### Payoffs

Using negative years in prison as utility gives

$$
\begin{aligned}
u_1(D,C)&=0, & u_1(C,D)&=-3, & u_1(D,D)&=-2, & u_1(C,C)&=-1,\\

\end{aligned}

Here $u_i(a_1,a_2)$ is Player $i$'s **payoff** for the action profile $(a_1,a_2)$. Higher utility means a better outcome.

The same sentence can be represented in equivalent ways:

| Quantity shown | Interpretation | Player's objective | Value for a 1-year sentence |
|---|---|---|---:|
| $t_i$ | Years in prison: a cost | Minimise | $1$ |

All three give the same ranking of outcomes. This chapter uses $u_i=-t_i$. Strictly, a payoff or utility is maximised; prison time itself is a cost to minimise.

### Payoff matrix

Player 1 chooses a row and Player 2 chooses a column; each cell gives the ordered payoff pair $(u_1,u_2)$.

$$
\begin{array}{c|c|c|}

\hline

$$

This is a **symmetric game**: swapping the prisoners' roles swaps their payoffs without changing the incentives. Symmetry does not require equal payoffs when they choose different actions.

### Assumptions for the one-shot analysis

We assume each prisoner is a **rational agent** who knows the actions and payoffs and maximises expected payoff. Each also has **self-regarding preferences**: only their own sentence enters their payoff. They cannot communicate or make enforceable agreements, and there are no later rounds for reward or retaliation.

## One-shot prisoner’s dilemma


### Best responses

A **best response** is the action that gives a player the highest payoff after the other player's action is held fixed.

- For Player 1, fix Player 2's column and compare the first number down that column.
- For Player 2, fix Player 1's row and compare the second number across that row.

The larger number is better because these payoffs are negative years in prison.

### Reading best responses from the matrix

Hold the other player's action fixed, then compare only the payoff entries belonging to the player whose decision you are analysing. This means moving across one row for Player 2 or down one column for Player 1—not comparing all four cells at once.

### Read across a row: Player 2

- If Player 1 chooses $C$: $0>-1$, so Player 2 chooses $D$.

### Dominance and Nash equilibrium

An action is a **strictly dominant strategy** when it gives a higher payoff than every alternative for every action of the other player. An action is a **strictly dominated strategy** when another action gives a higher payoff against every possible action of the other player. Here $D$ strictly dominates $C$ for both players.

A **unilateral deviation** changes only one player's action while holding the other player's action fixed. A **Nash equilibrium** is an action profile where no player can improve their payoff by such a change. At $(D,D)$, either player who switches alone moves from $-2$ to $-3$, so $(D,D)$ is the Nash equilibrium.

“Unilaterally” means that a player changes only their own action while the actions of everyone else remain fixed.

After a Nash equilibrium has been played, no player could obtain a higher payoff by changing their action alone.


Let $A_1$ and $A_2$ be the players’ action sets. An action profile $(a_1^*,a_2^*)$ is a Nash equilibrium when

$$
u_1(a_1^*,a_2^*) \geq u_1(a_1,a_2^*)
\qquad \text{for every } a_1\in A_1
$$

and


$$
u_2(a_1^*,a_2^*) \geq u_2(a_1^*,a_2)
\qquad \text{for every } a_2\in A_2
$$


An action profile $(a_1^*,a_2^*)$ is a Nash equilibrium when both inequalities hold.

<!-- reader-equation: b091f920-71fa-423d-bcca-802cdfc18199:0 -->

<!-- reader-equation: b091f920-71fa-423d-bcca-802cdfc18199:1 -->

<p style="font-size:0.68em;color:#5B6780"><strong>Notation:</strong> $A_i$ is player $i$'s action set; $a_i^*$ is the proposed equilibrium action. $u_i(a_1,a_2)$ is player $i$'s payoff; each inequality compares it with every unilateral alternative while the other player stays fixed.</p>


An outcome fails to be a Nash equilibrium when at least one player can gain by deviating alone. The “greater than or equal to” also allows indifference. Equality gives no strict gain from staying and nothing to gain by moving away, so the outcome still qualifies.

The asterisk marks the equilibrium action being considered; it does not assert dynamical stability.

### Pareto comparison

An outcome is **Pareto-superior** to another if nobody is worse off and at least one player is better off. An outcome is **Pareto optimal** (or Pareto efficient) if no feasible outcome improves someone’s payoff without worsening another’s.

Mutual cooperation gives one year each, rather than two years each under mutual defection. It is Pareto-superior: both prefer it. It is also Pareto optimal among our feasible outcomes, but is not a Nash equilibrium because either prisoner can gain by defecting alone.

The analysed payoff plane collects the same comparisons into one picture. An open point has at least one outgoing arrow: a player can improve their payoff by changing action alone. The filled point has no outgoing improvement arrow.

- Reward: $R = u_1(C, C)$ (payoff to a cooperator when both cooperate)

### Prisoner’s dilemmas outside prison

These are simplified two-player, one-shot examples. Each assumes the payoff ordering in Equation (8). Check that ordering before calling a real situation a Prisoner's Dilemma.

| Example | Cooperate | Defect | Assumed payoffs from worst to best |
|---|---|---|---|
| Shared water | Keep to a withdrawal quota | Take extra water | S. You keep to the quota while the other overdraws, leaving you with shortages.<br>P. Both overdraw, reducing the available supply.<br>R. Both keep to the quota, securing an adequate supply.<br>T. Only you overdraw, gaining extra water without a penalty. |
| Arms race | Limit armaments | Build up armaments | S. You limit armaments while the other builds up, leaving you vulnerable.<br>P. Both build up, paying for a costly arms race.<br>R. Both limit armaments, maintaining security at lower cost.<br>T. Only you build up, gaining a military advantage. |
| Price competition | Keep the higher price | Offer a fixed discount | S. You keep the higher price while the other discounts, losing customers.<br>P. Both discount, making smaller profits.<br>R. Both keep the higher price, making larger profits.<br>T. Only you discount, gaining enough customers to make the largest profit. |
| Road use | Avoid a shared shortcut | Use the shortcut | S. You avoid it while the other uses it, making the longest journey.<br>P. Both use it, creating congestion.<br>R. Both avoid it, making quicker journeys.<br>T. Only you use it, making the quickest journey. |

For water, taking extra must benefit you whether the other keeps to the quota or overdraws. Yet both keeping to the quota must be better than both overdrawing. Enforced penalties can change these incentives. P is the payoff when both defect, not an externally imposed penalty.

For prices, the discount must remain profitable. For road use, the journey times must have the stated ordering. Congestion alone does not establish a Prisoner's Dilemma.

Under these assumptions, each player prefers defection whichever action the other chooses. Both defect and receive P, although mutual cooperation would give both the better payoff R.

## Comparing game types

These examples all use $P=\{1,2\}$ and two actions per player; their action sets and payoff functions distinguish them.

A **coordination game** rewards compatible choices; an **anti-coordination game** rewards different choices. In an **asymmetric game**, the players' roles, actions or payoff rules are not interchangeable.

- Stag Hunt: $A_1=A_2=\{\text{stag},\text{hare}\}$. Mutual stag pays most; hare is safer when the other player does not hunt stag.
- Chicken (or Hawk–Dove): $A_1=A_2=\{\text{escalate},\text{yield}\}$. Each prefers the other to yield; mutual escalation is costly.
- Matching pennies: $A_1=A_2=\{\text{heads},\text{tails}\}$. Player 1 prefers matching actions; Player 2 prefers different actions.

## Return to rock–paper–scissors: randomising actions

We can now apply the equilibrium test to our opening game. No fixed action pair is a Nash equilibrium: after a draw either player could switch to win; after a win–loss outcome the losing player could switch to win.

A **pure strategy** selects one action with certainty. A **mixed strategy** assigns probabilities to the available actions. A **mixed Nash equilibrium** is a pair of such probability distributions where neither player can improve their expected payoff by changing their own distribution alone.

Expected payoffs average numerical utilities, so the gaps between payoffs now matter. Ordinal ranks alone are insufficient.

If the opponent chooses each action with probability one third, any action wins one third of the time, loses one third and draws one third. Its expected payoff is therefore zero. Any mixture of those actions also gives zero. When both players use this uniform mixture, neither can gain by deviating alone. This is the mixed Nash equilibrium.

Randomising actions in this way is different from maintaining equal numbers of fixed rock, paper and scissors agents. We next specify a population model to make that distinction concrete.

## Randomising actions

A **mixed strategy** assigns probabilities to actions. A **mixed Nash equilibrium** is a profile of mixtures where no player can improve expected payoff by changing their mixture alone.

In rock–paper–scissors, each player chooses each action with probability one third. Against this mixture every action has expected payoff zero; no unilateral change earns more.

## From pairwise games to a moving population

The next animation adds a population process to the pairwise win–loss rule. Each agent carries a fixed type between conversions; it is not a player randomly choosing a fresh action at every encounter. This is a population of pairwise interactions, not one simultaneous game involving every agent.

The model rules are:

1. each agent carries one of the three types;
2. agents move through a finite domain;
3. nearby agents encounter one another;
4. the cyclic rule determines the local winner;
5. the defeated agent is converted to the winner's type.

### Population behaviour

The payoff matrix determines who defeats whom in one encounter. Population history also depends on movement, collision rate, spatial clustering, boundary conditions, update order, initial numbers and chance because these determine which encounters occur.

This also explains an apparently strange route to victory. Scissors cannot remove rock directly. For scissors to fill the arena, paper may first need to eliminate the rocks; scissors can then invade the remaining paper population. A type's prospects therefore depend on the abundance and spatial arrangement of all three types. This is **frequency-dependent selection**: the success of a type depends on the population mixture, here with **cyclic dominance**, where each type defeats one alternative but is defeated by another.

In a **well-mixed population**, encounter probabilities depend on the overall population mixture, not spatial neighbours. Some well-mixed deterministic models sustain cycles, but the long-run behaviour depends on the population update rule. Extinction in a finite agent simulation is a population-level outcome produced by chance differences in individual encounters and replacements and the implementation of space and interaction, not a new entry in the payoff matrix.

- **Selection** determines which strategies leave more descendants or copies.

- **Mutation** changes copied strategies or introduces new ones.

### Selection changes the population

1. Agents play and accumulate payoff.
2. Strategies with greater fitness are copied more often.
3. Copies inherit a strategy, with mutation if included.
4. The changed population supplies the opponents for later encounters.

### Evolution of strategy frequencies

A **strategy frequency** is the fraction of a population using that strategy. **Replicator dynamics** models how these fractions change when strategies earn different payoffs.

Let $x_s$ be the fraction of a **well-mixed population** using strategy $s$, where encounters are determined by the population mixture. Its expected payoff $\pi_s(\mathbf{x})$ depends on the current mixture $\mathbf{x}$. The population mean payoff is

$$
\bar\pi(\mathbf{x})=\sum_s x_s\pi_s(\mathbf{x}).
$$

The **replicator equation** is

$$
\dot{x}_s=x_s\left[\pi_s(\mathbf{x})-\bar\pi(\mathbf{x})\right].
$$

The rate of change equals the current share multiplied by its payoff advantage over the population mean. The dot denotes a continuous-time rate of change.

- A positive payoff advantage increases its share. A negative advantage decreases it. Zero means no change at that instant.
- An absent strategy stays absent because its current share is zero.
- The mean payoff is weighted by strategy frequencies. Subtracting it keeps the shares summing to one.

A strategy earning 3 against a population average of 2 increases its share. Its advantage can change as the population mixture changes.

The basic equation assumes an effectively infinite, well-mixed population, faithful inheritance and no mutation. Payoff differences determine relative per-capita growth rates. Negative payoffs are allowed because only payoff differences enter. Sampling descendants in a finite population adds random variation. See [Cressman and Tao (2014)](https://pmc.ncbi.nlm.nih.gov/articles/PMC4113915/) for the mathematical background.

In a **structured population**, a lattice or network restricts encounters to neighbours. Strategies can spread through local copying. Cooperative clusters can protect their members, and cyclic competition can produce fronts and spirals.
