# What is a game?

Game theory represents strategic interaction: players choose actions, and each player’s payoff depends on the combination of choices. A one-shot game is specified by

$$
G=\left(P,\{A_i\}_{i=1}^{n},\{u_i\}_{i=1}^{n}\right).
$$

Here $P$ is the **player set**. The subscript identifies which player is meant: $A_i$ is player $i$’s **action set**, and $u_i$ is player $i$’s **payoff function**. An **action profile** lists one chosen action for every player, and the payoff function evaluates the resulting profile for each player. This is the standard normal-form representation of a one-shot game.

A **payoff** is the numerical value assigned to an outcome; **utility** is the same quantity here (the broader decision-theory term). An **action** is a choice available in the current move. A **strategy** is a rule for choosing actions: in a one-shot game it selects one action, while in repeated play it can depend on the earlier history.

A standard assumption is that players are **rational agents**: they choose an available action that maximises their payoff given their information and beliefs. Under uncertainty, this means maximising expected payoff. This is a modelling choice about the decision rule, not a claim that people always calculate perfectly.

## Payoff matrices

For two players, a **normal-form payoff matrix** (often called a **payoff table**) lists Player 1’s actions in rows and Player 2’s actions in columns. Read a row as Player 1’s action and a column as Player 2’s action. Each cell gives the ordered pair

$$
(\text{Player 1 payoff},\;\text{Player 2 payoff})
$$

For two actions per player, write Player 1’s actions as $\alpha_1,\alpha_2$ and Player 2’s as $\beta_1,\beta_2$. The generic payoff matrix is

$$
\begin{array}{c|c|c|}
 & \text{Player 2: } \beta_1 & \text{Player 2: } \beta_2 \\
\hline
\text{Player 1: } \alpha_1 & (u_1(\alpha_1,\beta_1), u_2(\alpha_1,\beta_1)) & (u_1(\alpha_1,\beta_2), u_2(\alpha_1,\beta_2)) \\
\hline
\text{Player 1: } \alpha_2 & (u_1(\alpha_2,\beta_1), u_2(\alpha_2,\beta_1)) & (u_1(\alpha_2,\beta_2), u_2(\alpha_2,\beta_2)) \\
\hline
\end{array}
$$

This is the one-shot representation we will use before adding repetition, movement or population structure.

For a fixed action by the other player, compare the payoff entries for the player whose action is being analysed. The action with the larger payoff is that player’s **best response**. Preferences are represented by the payoff function, so social preferences require a richer payoff function.



## Familiar example: rock–paper–scissors

Here two players simultaneously choose rock, paper or scissors without seeing the other’s choice. Rock beats scissors, scissors beats paper, and paper beats rock; matching actions draw. Give a win payoff (+1), a loss payoff (-1), and a draw payoff (0).

It is also a **symmetric game**: exchanging the players’ roles exchanges their payoffs without changing the rules. It has **cyclic dominance**: no action is best against everything, because each defeats one alternative and is defeated by another.

### A biological rock–paper–scissors game

Male side-blotched lizards (*Uta stansburiana*) have three throat-colour types, or morphs, associated with different mating strategies. In the population studied by [Sinervo and Lively (1996)](https://doi.org/10.1038/380240a0):

- Orange males defend large territories and displace blue males.
- Blue males guard mates closely and prevent yellow males from sneaking matings.
- Yellow males mimic females and sneak matings within orange males’ territories.

This gives the cycle:

| Pair of morphs | Higher expected mating success | Why |
|---|---|---|
| Orange and blue | Orange | Orange males can displace blue males. |
| Blue and yellow | Blue | Blue males guard mates from yellow males. |
| Yellow and orange | Yellow | Yellow males can sneak matings in orange territories. |

Here “wins” means greater expected reproductive success in that matchup, not that one lizard wins every encounter. The payoff is the number of offspring sired, and it depends on which morphs are common. The simplified cycle is orange → blue → yellow → orange, so no morph has an advantage against all the others.

| Game element | Biological meaning |
|---|---|
| **Players** | Competing male side-blotched lizards |
| **Strategies** | Territory defence, mate guarding or sneaking, associated with throat-colour morphs |
| **Payoff** | Number of offspring sired, depending on the other males' strategies |

# Modelling games

A game is a representation of strategic decisions: the outcome for each player depends on what all players do.

A **rational agent** has consistent preferences over outcomes and chooses an available action that maximises their payoff, given their information and beliefs about the other players. When outcomes are uncertain, we assume the agent maximises **expected payoff**: the average payoff weighted by the probabilities they assign to those outcomes. Payoffs represent what the agent values.

For example, in rock–paper–scissors, paper is rational if the opponent is certain to choose rock. Rationality does not require knowing the opponent’s action in advance: the agent maximises expected payoff using their beliefs. The agent remains motivated by the payoff specified in the model. Representing altruism, fairness or concern for others requires a richer payoff function and is an additional modelling choice.

A one-shot strategic game is specified by

$$
G=\left(P,\{A_i\}_{i=1}^{n},\{u_i\}_{i=1}^{n}\right).
$$

Here $P=\{1,\ldots,n\}$ is the **player set**, and an **action set** $A_i$ lists player $i$’s available actions. The action sets determine $S=A_1\times\cdots\times A_n$; each member is an **action profile** $\mathbf{a}=(a_1,\ldots,a_n).$ Each **payoff function** $u_i:S\to\mathbb{R}$ assigns player $i$ a utility for every profile.

A **rational agent** chooses the available action with the highest payoff, given their information and beliefs. Under uncertainty, they maximise **expected payoff**.

In rock–paper–scissors: if the opponent is certain to choose rock, choose paper.

The model's payoff function defines the objective. **Self-regarding preferences** use only the agent's outcome; social preferences require a richer payoff function.


Rock–paper–scissors therefore instantiates Equation (1) as $P=\{1,2\}$ and $A_1=A_2=\{\text{rock},\text{paper},\text{scissors}\}.$ The payoff table specifies $u_1$ and $u_2$ for every action profile; for example, $u_1(\text{rock},\text{paper})=-1$ and $u_2(\text{rock},\text{paper})=+1.$

Equation (1) specifies incentives. Population structure, learning, movement and repeated interaction are additional modelling choices that determine how those incentives unfold through time.


## Payoff matrices

A **payoff matrix** displays both players’ payoffs for every action pair. This is the **normal form** of the game: players, available strategies and their payoffs, without a sequence-of-moves diagram.

Here Player 1’s two actions are $\alpha_1,\alpha_2$ and Player 2’s are $\beta_1,\beta_2.$ The subscripts distinguish actions, not players. Rows select Player 1’s action, columns select Player 2’s, and each entry lists Player 1’s payoff first:


$$
\begin{array}{c|c|c|}
 & \text{Player 2: } \beta_1 & \text{Player 2: } \beta_2 \\
\hline
\text{Player 1: } \alpha_1 & (u_1(\alpha_1,\beta_1), u_2(\alpha_1,\beta_1)) 
                           & (u_1(\alpha_1,\beta_2), u_2(\alpha_1,\beta_2)) \\
\hline
\text{Player 1: } \alpha_2 & (u_1(\alpha_2,\beta_1), u_2(\alpha_2,\beta_1)) 
                           & (u_1(\alpha_2,\beta_2), u_2(\alpha_2,\beta_2)) \\
\hline
\end{array}
$$

- Player 1’s actions label the rows.
- Player 2’s actions label the columns.
- Each cell of the matrix corresponds to one strategy profile $s\in S$, and the entries show the utilities $(u_1,u_2)$ corresponding to that action pair.

This compact form is called the normal form of a game.

# Canonical model: Prisoner’s dilemma

Rock–paper–scissors gives us a familiar zero-sum contest. The Prisoner’s Dilemma will show us how individual incentives can pull against an outcome that would benefit both players.


## Canonical model: Prisoner’s dilemma

We now use the Prisoner’s Dilemma as the canonical two-player model: individual incentives can pull against an outcome that would benefit both players.


## Prisoner's dilemma

<p style="margin:0.15em 0 0.75em;color:#376FA8;font-style:italic;line-height:1.18"><strong>Scenario:</strong> Two suspects are questioned separately after a burglary. Both stay silent and face only a trespassing charge; if one testifies, they provide evidence of the burglary and the other faces the more serious charge. Each sentence depends on both choices.</p>
<p style="margin:0.25em 0 0.35em"><strong>Model:</strong> A one-shot game gives each prisoner one choice; there is no communication or enforceable agreement, and no later round for retaliation or reward.</p>
<p style="margin:0.25em 0 0.35em"><strong>Pure strategy:</strong> choose one action with certainty.</p>
<p style="margin:0.25em 0 0.35em"><strong>Mixed strategy:</strong> randomise over actions, for example testify with probability <em>p</em> and stay silent with probability <em>1 − p</em>.</p>
<p class="small-note">In a repeated game, a strategy can depend on observed history, such as Tit for Tat. Here each prisoner’s utility records their own sentence.</p>

## Sentence outcomes

| What happens | Sentence |
|---|---|
| One testifies; the other stays silent | The testifying suspect goes free; the silent suspect serves 3 years |
| Both testify | 2 years each |
| Both stay silent | 1 year each |

Testifying is individually tempting, although mutual silence is better for both. Now represent this situation as a game.

### Set up the game

| Component | Specification |
|---|---|
| **Players** | $P=\{1,2\}$ |
| **Actions** | $A_1=A_2=\{C,D\}$; $C$ means stay silent and $D$ means testify |
| **Payoffs (utilities)** | $u_1$ and $u_2$ are negative prison sentences; less negative is better |

<p style="font-size:0.68em;color:#5B6780"><strong>Notation:</strong> Subscripts identify the player: <em>A</em><sub>1</sub> and <em>u</em><sub>1</sub> belong to Player 1; <em>A</em><sub>2</sub> and <em>u</em><sub>2</sub> to Player 2. Here <em>C</em> means cooperate (stay silent) and <em>D</em> means defect (testify). Other sources may reverse these labels: <em>C</em> can mean confess (testify) and <em>D</em> can mean do not confess (stay silent). Always check the action definitions and how utility is interpreted: are larger values better, or is a cost being minimised?</p>


The set of **action profiles** contains all possible action pairs, and for each profile in $S$, the payoff functions $u_1$ and $u_2$ give the two prisoners' utilities:

$$
S=\{C,D\}\times\{C,D\}=\{(C,C),(C,D),(D,C),(D,D)\}
$$

The set of **action profiles** contains all possible action pairs.

$$
\begin{aligned}

“Unilaterally” means that a player changes only their own action while the actions of everyone else remain fixed. After a Nash equilibrium has been played, no player could obtain a higher payoff by changing their action alone.

Let $A_1$ and $A_2$ be the players’ action sets. An action profile $(a_1^*,a_2^*)$ is a Nash equilibrium when

$$
u_1(a_1^*,a_2^*) \geq u_1(a_1,a_2^*)
\qquad \text{for every } a_1\in A_1
$$

and


$$
u_2(a_1^*,a_2^*) \geq u_2(a_1^*,a_2)
\qquad \text{for every } a_2\in A_2
$$

An outcome fails to be a Nash equilibrium when at least one player can gain by deviating alone. The “greater than or equal to” also allows indifference. Equality gives no strict gain from staying and nothing to gain by moving away, so the outcome still qualifies.

The asterisk marks the equilibrium action being considered. It indicates stability against a unilateral payoff-improving change, not dynamical stability: that would require an update rule and a time-evolving state.

#### Pareto comparison

An outcome $x$ is **Pareto-superior** to a feasible outcome $y$ if every player’s payoff at $x$ is at least as high as at $y$, and at least one player’s payoff is higher. An outcome is **Pareto optimal** (or Pareto efficient) if there is no other feasible outcome that is Pareto-superior to it.

In the Prisoner’s Dilemma, mutual cooperation gives one year each, rather than two years each under mutual defection. It is Pareto-superior: both prefer it. It is also Pareto optimal among our feasible outcomes, but is not a Nash equilibrium because either prisoner can gain by defecting alone.

The payoff plane shows these same comparisons geometrically. An open point has at least one outgoing arrow: a player can improve their payoff by changing action alone. The filled point has no outgoing improvement arrow.

<p class="small-note"><strong>Notation:</strong> $T$ is temptation (defect against cooperation); $R$ mutual-cooperation reward; $P$ mutual-defection punishment; $S$ the cooperator's payoff against defection. Larger utility is better.</p>


For Player 1 (and, by symmetry, Player 2), define the possible utilities as:
- Reward: $R = u_1(C, C)$ (payoff to a cooperator when both cooperate) 
- Temptation: $T = u_1(D, C)$ (payoff to a defector against a cooperator) 
- Punishment: $P = u_1(D, D)$ (payoff to a defector when both defect)
- Sucker’s payoff: $S = u_1(C, D)$ (payoff to a cooperator against a defector)


Let $T$, $R$, $P$ and $S$ denote the temptation, reward, punishment and sucker's payoffs. The Prisoner's Dilemma condition is

$$
T>R>P>S
$$

- $T>R$ and $P>S$: defection is dominant.
- $R>P$: mutual cooperation is better for both than mutual defection.

**Ordinal payoffs** rank outcomes without measuring the gaps between them. Equation (8) gives this ordering. We can replace the four payoffs by ranks 1 to 4 without changing the pure-action best responses.

These are utilities: larger values are preferred. If the original measurement is a cost such as prison years, convert it to utility before applying this ordering.

For the prison-sentence matrix used above, $\left(T,R,P,S\right)=(0,-1,-2,-3)$ because utility is negative years in prison. The Axelrod tournament uses the conventional positive scores $\left(T,R,P,S\right)=(5,3,1,0).$ The ordering is the same, but the gaps differ; this is a deliberate tournament convention because cumulative scores depend on those gaps. The evolutionary extension uses the same tournament scores, so raw score magnitudes should not be compared across the two conventions.

The distinction matters once play is repeated. The ordinal ordering identifies the Prisoner’s Dilemma, but tournament payoffs are cardinal: round payoffs are added, so the gap sizes affect cumulative scores. In particular, the gain from exploiting cooperation, the loss suffered by the cooperator, and the size of the punishment after defection determine how costly retaliation is and how quickly a one-round advantage is offset.

### Applications beyond prison

These simplified two-player, one-shot examples illustrate where the Prisoner’s Dilemma ordering might apply. Each assumes the ordinal payoff ranking in Equation (8).

| Example | Cooperate | Defect | Assumed payoffs from worst to best |
|---|---|---|---|
| **Shared water** | Keep to a withdrawal quota | Take extra water | $S.$ You keep to the quota while the other overdraws, leaving you with shortages.<br>$P.$ Both overdraw, reducing the available supply.<br>$R.$ Both keep to the quota, securing an adequate supply.<br>$T.$ Only you overdraw, gaining extra water without a penalty. |
| **Arms race** | Limit weapons | Build up weapons | $S.$ You limit weapons while the other builds up, leaving you vulnerable.<br>$P.$ Both build up, paying for a costly arms race.<br>$R.$ Both limit weapons, maintaining security at lower cost.<br>$T.$ Only you build up, gaining a military advantage. |

## Repeated play and evolving strategy

A repeated game and an evolutionary population are different extensions of a pairwise game. In repeated play, a strategy can respond to earlier actions but its rule stays fixed during the tournament. In an evolutionary model, payoff-dependent copying or reproduction changes the strategy frequencies, so later encounters come from a different population.

We first compare fixed strategies in an Axelrod-style tournament. We then use those tournament scores to introduce payoff-dependent population updating, before looking at two other population models: copying by successful neighbours in a static population, and conversion after a loss in a moving population.




## Repeated play and evolving strategy

A repeated game can use history while the strategy rule stays fixed. Evolutionary game dynamics changes strategy frequencies through payoff-dependent copying or reproduction.

We compare fixed strategies first, then examine strategy updating in static and moving populations, and finally let tournament strategies change in a population model.


### In a heterogeneous population: Axelrod Prisoner’s Dilemma tournament

In a **repeated game**, the same two players meet for several rounds. A strategy can use earlier actions to choose the next one. The score for a match is the sum of its round payoffs (or their discounted sum if later rounds count less).

Compared with the classical one-shot model, the Axelrod tournament loosens the no-history constraint: memory makes retaliation, forgiveness and reputation-like responses possible. It still assumes no direct communication and no enforceable agreement. It also changes the choice question: the strategies are fixed rules, not players recomputing a one-shot best response or copying one another during a match. The tournament compares how those rules perform against a declared opponent field.

An Axelrod-style tournament keeps the strategy rules fixed and compares them over many opponents. For each pair of strategies:

1. play repeated matches;
2. add the round payoffs within each match;
3. average over repeated matches;
4. average across the declared opponent field and rank the strategies.

This is tournament performance, not evolution: no strategy is copied, removed or introduced. The ranking depends on the strategies entered, the match length, the payoff matrix, errors and how opponents are weighted. If match lengths differ, report mean payoff per round rather than cumulative score.

For Figure 10.7, a **field** is the declared collection of opponent strategy types and their sampling weights. A tournament is defined as follows: one round is one simultaneous Prisoner’s Dilemma encounter. One repeated match is 50 rounds between the same focal strategy and opponent; histories reset before the next match. A tournament is 300 such matches for each focal strategy, with one opponent sampled uniformly with replacement from the field before each match. The x-axis counts completed tournament matches, *m*. At vertical position *m*, each line is the focal strategy’s running mean payoff per round across the first *m* matches; moving one step right adds one newly sampled opponent and recomputes that cumulative average. Heatmap cells average 30 independent 50-round matches for one focal/opponent pair, so a long-run tournament line approaches the row-average of the heatmap under uniform sampling, not one particular cell. Strategies remain fixed: this is tournament analysis, not evolution.

#### Properties of successful strategies

Later analyses highlighted four properties often found in strong tournament strategies:

| Property | Behaviour in the tournament | Trade-off |
|---|---|---|
| **Nice** | Cooperate on the first move; do not exploit an unprovoked opponent. | Can be exploited by persistent defectors. |
| **Retaliatory** | Respond to defection rather than allowing repeated exploitation. | Excessive punishment can destroy future cooperation. |
| **Forgiving** | Resume cooperation after the opponent returns to cooperation. | Too much forgiveness can invite repeated exploitation. |
| **Non-envious** | Seek a good score rather than needing to outscore the opponent in every match. | The ranking depends on the opponent field, horizon and noise. |

Tit for Tat has the four properties in their simplest form: it starts cooperatively, answers defection, returns to cooperation when the opponent does, and does not try to beat a cooperative opponent. Its tournament score still depends on the opponents, match length and noise.


#### Tournament scores can feed an evolutionary model

The tournament keeps the four strategy rules fixed. To turn its scores into an evolutionary model, use payoff as a signal of copying or reproductive success. Figure 10.8 uses the same four strategies in a finite, well-mixed population of 2,000 agents, starting in equal shares. One round is one Prisoner’s Dilemma encounter; 50 rounds make one repeated match. For each pair of strategies, 80 independent matches are used to estimate the mean payoff per round. The 80 is the ensemble size, not another time scale.

An agent keeps its strategy throughout a repeated game; it does not jump to another strategy after an individual round. Between generations, strategies with higher mean payoff are copied more often, using selection strength 0.08 and no mutation. **Selection** maps payoff to copying success, determining which strategies are copied more often. **Inheritance** performs that copying by passing the selected strategy into the next generation, and **mutation** allows copies to change. These are additional model choices. The coloured bands show population shares, not individual actions. In this run Tit for Tat becomes the largest share, but that outcome depends on the opponent field, payoff ordering, match length and update rule.

A tournament scores strategies against a fixed collection of opponents.

### From payoff to fitness

**Fitness** means expected reproductive or copying success. We specify how payoff affects fitness.

- **Selection** determines which strategies leave more descendants or copies.
- **Inheritance** copies the selected strategy into descendants or the next generation.
- **Mutation** changes copied strategies or introduces new ones.

### Selection changes the population

1. Agents play and accumulate payoff.
2. Strategies with greater fitness are copied more often.
3. Copies inherit a strategy, with mutation if included.
4. The changed population supplies the opponents for later encounters.

### Evolution in action

The tournament and this model both contain populations. In the tournament, the strategy set and its frequencies stay fixed. Here, agents play in random pairs, then some adopt higher-payoff strategies for the next generation, so the mix changes. This is population updating, not copying an opponent’s action within a round. In the Prisoner’s Dilemma, Defect rises from 0.50 toward 1.
