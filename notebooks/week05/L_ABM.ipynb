# Agent-based modelling

<div class="canonical-model-marker"><span>Canonical model</span><strong>Vicsek flocking model</strong></div>

<div class="modelling-practice-marker"><span>Modelling practice</span><strong>Design and refine parameter sweeps</strong></div>

<div class="toolkit-addition-marker"><span>Toolkit additions</span><strong>Mobile agents, vector headings, local interaction and stochastic updates</strong></div>



Agent-based modelling is useful when individuals remain distinguishable and their locations, attributes, decisions, or interaction partners matter. An agent has agency in the modelling sense: its next action depends on its own state and locally available information. In this sense, an agent is situated: what it can observe and do depends on the world it occupies. This need not imply conscious choice. A self-propelled particle has an effectively available energy supply, whereas a person, animal or robot may follow goals and hierarchical decision rules.

The modelling question determines whether an agent-based description is useful and whether to represent individual variation or average it away.

This week uses collective motion to define agents and their environment, inspect one update, choose a system-level observable, and design a parameter sweep. Noise is the main investigation: we first sweep it coarsely, repeat each condition, and then concentrate new simulations where the collective response changes. Density provides a second dimension, while a final speed screen shows how to justify treating a parameter as approximately unimportant only over a stated range.

Franklin's 1000th goal establishes the modelling problem: coordinated system-level motion can arise without a central controller, while congestion, social cues, goals and boundaries still matter.

Self-organisation occurs when interactions among lower-level components produce a global spatial, temporal or functional pattern without a central controller specifying that pattern. External driving and constraints may still be essential: the crowd has a field and a milestone; a flock has an environment; active matter consumes energy.

At the microscopic level we could record positions, headings, neighbours and decisions. At the macroscopic level we see a surge, a stream or a jam. An agent-based model keeps enough of the first description to ask how the second emerges.

### Before local-interaction models

The coordination of a large flock once seemed to require communication across the whole group. Several explanations were considered:

- Leader or command: the most immediate analogy was command-and-control, in which a leading bird signals and the others obey. Edmund Selous described this as the apparently simplest explanation, but his observations of near-simultaneous turns made a single commander increasingly difficult to identify.
- Thought transference: Selous eventually proposed that some form of thought transfer might coordinate a flock. His 1931 book *Thought-Transference (or What?) in Birds* documents careful observations, but its proposed mechanism has not survived.
- A threshold signal: Davis (1980) proposed that preliminary movements by enough birds could announce an imminent turn.
- A propagating manoeuvre: film analysis by Potts (1984) showed that one bird can initiate a turn that travels through the flock as a wave. Each bird responds locally as the wave reaches it, more like a stadium wave than a military order.

The important change was from asking how every bird receives one global instruction to asking how information propagates through local interactions. Modern models test whether those local rules are sufficient; observations are still needed to determine which rules real birds use.

<p class="media-credit">Sources: Selous (1931), <em>Thought-Transference (or What?) in Birds</em>; Potts (1984), <a href="https://doi.org/10.1038/309344a0">“The chorus-line hypothesis of manoeuvre coordination in avian flocks”</a>, <em>Nature</em> 309, 344–345.</p>


Mathematically, “flocking” is used more broadly than its biological collective noun. It can describe coherent local motion in fish, bacteria, cells, robots and synthetic active matter. The shared observation is ordered movement; the mechanism must still be established for each system.

Native budgerigars provide a local example in which collective motion is coupled to changing environmental conditions.

The question determines what the model represents and how it should be analysed. The same words can also shift meaning between disciplines: an animator, physicist and behavioural biologist may use *state*, *interaction* and *success* for different quantities and standards of evidence. We use Vicsek because it isolates alignment and noise most simply.

### The Heisenberg comparison

The classical Heisenberg model places a three-component unit spin on every fixed lattice site. Neighbouring spins interact through their dot product, so ferromagnetic coupling favours alignment. The two-component XY model is the planar version: each fixed spin is described by an angle.

Vicsek borrowed the alignment idea and changed the physical problem. The headings are planar and therefore XY-like, the particles move through space, neighbours continually change, noise is imposed dynamically, and energy is continually supplied through self-propulsion. The result is a driven, non-equilibrium active-matter model. The XY model provides a conceptual bridge rather than a mathematical identity.

## Build the Vicsek model

An agent-based model is a framework rather than one fixed equation. It specifies what each agent stores, how agents interact, how they update and what is recorded. Those parts can be replaced independently, so two agent-based models may represent quite different mechanisms.

Vicsek and colleagues asked what happens when self-propelled particles align with nearby particles while noise perturbs their headings.

- State: positions, headings, local neighbours and noise.
- Remove: aerodynamics, body shape, vision, memory, leadership and attraction.

The model isolates local alignment as a possible mechanism for collective motion. Later variants replace the neighbourhood, motion, interaction or noise rule one at a time; the agent-based framework remains.

### Averaging headings with complex exponentials

Angles wrap around: the ordinary arithmetic mean of $1^\circ$ and $359^\circ$ is $180^\circ$, although both headings point almost due east. Map each heading to the unit vector

$$
e^{\mathrm i\theta_j}=\cos\theta_j+\mathrm i\sin\theta_j
$$

Adding these vectors averages their horizontal and vertical components. The updated direction can therefore be written directly as

$$
\bar\theta_i(t)=\operatorname{Arg}\!\left(\sum_{j\in\mathcal N_i(t)}e^{\mathrm i\theta_j(t)}\right)
$$

Here $\theta_j$ is neighbour $j$'s direction and $\bar\theta_i$ is the mean neighbour direction used to update agent $i$. The sum is a temporary resultant vector, not a second direction variable. Dividing it by the number of neighbours would change its length but not its direction. If the vectors cancel exactly, the resultant is zero and there is no unique mean heading; an implementation must decide how to handle that rare case.


## Angular noise

$$
\theta_i'=\bar\theta_i+\xi_i,
\qquad \xi_i\sim U[-\eta/2,\eta/2]
$$

This is the noise convention used in the original Vicsek model. The agent first estimates the local mean heading $\bar\theta_i$, then makes a turning error $\xi_i$. The parameter $\eta$ controls the width of the error distribution.

**Interpretation:** uncertainty enters while the agent executes a chosen direction.

## Parameter sweep

<img src="images/vicsek_parameter_sweep_design.svg" alt="A coarse noise sweep refined near its steepest change, a density sweep, and a two-dimensional noise-density map of Vicsek polarisation" style="display:block;width:100%;max-height:430px;margin:0 auto">

<p class="figure-caption" id="fig-w5-2"><strong>Figure 5.2.</strong> Parameter sweeps produced by our angular-noise Vicsek implementation. Left: a coarse noise sweep at fixed density, followed by additional points across the steepest interval and its more strongly changing neighbour. Centre: a density sweep at fixed noise. Right: a coarse two-parameter map; the white lines mark the one-dimensional slices, and the orange contour marks Φ = 0.5. One-dimensional points summarise 10 independent runs; map cells summarise six.</p>


A parameter sweep is a controlled computational experiment whose sampling design specifies where the parameter space is evaluated. Vicsek et al. varied noise at fixed density and density at fixed noise. Figure 5.2 qualitatively recreates those two experiments with the model defined above, then combines their control parameters. It is not intended to reproduce every numerical value in the published figure.

Before running a sweep, specify the varied parameter, the quantities held fixed, the observable, the settling time, the averaging window and the random repetitions. Here $N=120$, $R=\Delta t=1$ and $v=0.1$. We vary density by changing the box width using $L=\sqrt{N/\rho}$. Each run lasts 300 updates and is represented by mean polarisation over its final 80 updates.

The first panel begins with nine noise values across the full meaningful angular-noise range. Its mean response changes appreciably across two neighbouring intervals, so both are sampled more densely using new random runs. Selecting the steepest interval plus its more strongly changing neighbour resolves the broader transition band—including the change between noise widths 2 and 3—rather than treating one coarse pair as the whole transition. This spends computation where it can resolve the response rather than placing equally many points in regions that are already flat. A large change, intermediate response or unusually large run-to-run variation can all motivate refinement.

The exact refinement interval is conditional on the pilot experiment. This Reader example uses density ρ = 2, N = 120, nine coarse noise values and 300 updates. The workshop uses density ρ = 0.8, N = 80, seven coarse values and 180 updates, so its loss of order begins at weaker noise. Both apply the same refinement rule, but they should not be expected to select identical numerical intervals. This difference previews the density dependence examined later in the sweep.

The right panel treats the one-dimensional experiments as slices through the same two-parameter space. Increasing density allows alignment to survive stronger noise, so the transition region forms a boundary rather than occurring at one universal noise value. The map uses a deliberately coarse grid: smooth interpolation would imply precision that was not simulated.


### Sweep design

1. Start with a coarse grid covering the plausible range.
2. Repeat each condition and summarise the response.
3. Refine where the response changes rapidly or becomes especially variable.
4. Keep the pilot points visible and state what remained fixed.

A uniformly fine grid can waste computation in flat regions; a fine grid with one run per point can mistake stochastic variation for parameter structure.

### Testing whether a parameter matters

<img src="images/vicsek_speed_screen.svg" alt="Vicsek polarisation against displacement per update for ordered, near-transition and disordered conditions" style="display:block;width:100%;max-width:900px;margin:1rem auto">

<p class="figure-caption" id="fig-w5-speed-screen"><strong>Figure 5.3.</strong> A speed-screening experiment at fixed density ρ = 2. Lines show mean late-time polarisation and bands show the middle 50% across 12 independent runs. Speed is expressed as displacement per update relative to interaction radius, <i>v</i>Δ<i>t</i>/<i>R</i>. The noise settings represent clearly ordered, near-transition and clearly disordered conditions.</p>

Vicsek et al. fixed $R=\Delta t=1$ and reported that their results were insensitive to speed over $0.003<v<0.3$. A sweep can test whether that simplification is adequate for our question. Figure 5.3 supports it well away from the transition: changing speed barely changes whether the system is clearly ordered or disordered. Near the transition, however, the response shifts by more than the run-to-run spread. Slow motion also renews the neighbour network slowly, so a fixed number of updates may not represent an equally long physical exploration at every speed.

The defensible conclusion is conditional: speed can be held fixed for conclusions that are robust across the tested range, but it has not been removed from the model. A parameter screen must name its range, observable, other parameter settings and simulation duration. The limits $v\to0$ and very large $v$ are qualitatively different.

## Noise and transition order

Vicsek et al. (1995) reported a continuous transition: as noise increased, the polarisation appeared to fall smoothly to zero, with behaviour reminiscent of a continuous critical point. This made the minimal model a candidate for a new nonequilibrium universality class.

Grégoire and Chaté (2004), followed by larger simulations by Chaté et al. (2008), reported discontinuous, first-order-like behaviour. Near the transition, ordered high-density travelling bands can coexist with a disordered low-density background. In systems too small to accommodate the bands, or runs too short to sample their appearance, the change can look deceptively smooth.

The alternatives describe different mechanisms. A continuous transition is organised by growing fluctuations and an increasing correlation length; critical exponents and universality become central. A discontinuous transition involves a jump and coexistence between ordered and disordered states; nucleation, hysteresis, intermittency and travelling density bands become central. The disagreement concerned the physical phenomenon underlying the graph. This is a transition between macroscopic collective states; classical chaos is a separate question.

The location of the noise also matters. With angular noise, each agent first computes a local mean direction and then makes a turning error. With vectorial noise, a random vector perturbs the alignment signal before its direction is taken. These are different microscopic models, not two interchangeable ways of coding one equation. Vectorial noise made discontinuity easier to see; the angular-noise case required much larger length and time scales before banding and discontinuity became clear.

Diagnosing a continuous transition requires more than a smooth curve from one convenient simulation size. Noise convention, density, speed, boundaries, initialisation, run length and system size belong in the claim. Useful evidence includes morphology, order-parameter distributions, hysteresis, Binder cumulants and finite-size scaling.

If the transition were genuinely discontinuous, Vicsek's log--log construction would not remain a straight line all the way to the transition. Writing $\epsilon=(\eta_c-\eta)/\eta_c$, a continuous transition has $\Phi\sim\epsilon^\beta$ and therefore $\log\Phi=\beta\log\epsilon+\text{constant}$. At a discontinuous transition, $\Phi$ approaches a non-zero jump $\Phi_0$ as $\epsilon\to0^+$, so $\log\Phi$ bends towards the horizontal value $\log\Phi_0$ rather than continuing down a line. Over a restricted finite-size range, however, the rounded curve can still look approximately straight. Chaté et al. therefore tested discontinuity using distributions, Binder cumulants, switching time series and finite-size scaling rather than fitting another critical exponent.

### Finite-size scaling

<div class="optional-reader-flag"><strong>Optional · mathematical detail</strong> The formal scaling ansatz is not directly assessable. The practical point remains useful: a finite system can round or shift an apparent transition.</div>

A finite simulation rounds a transition and can shift its apparent location because it cannot contain correlations or structures larger than its domain. In an idealised statistical-physics model, finite-size scaling compares increasing $L$ values to infer whether a transition survives as $L\to\infty$. One system size cannot establish the order of a transition.

A common continuous-transition hypothesis is

$$
\Phi(\eta,L)=L^{-\beta/\nu}\,\mathcal F\!\left((\eta-\eta_c)L^{1/\nu}\right)
$$

The practical message is that the order parameter depends on both the control parameter and system size. Useful evidence includes how the transition location, rounding, order-parameter distribution, switching behaviour and Binder cumulant change with $L$.

For an empirical complex system, finite size is often part of the phenomenon rather than an error to remove. A flock of 20 animals may genuinely behave differently from one of 20,000 because boundaries, communication times, subgroup formation and rare events scale differently. The relevant question is then which behaviours depend on size and which remain comparable across sizes. The infinite-system limit can still provide a benchmark, but it is not automatically the true system that reality approximates.

The comparison also requires a clear control. Increasing $N$ while holding $L$ fixed changes density. Increasing $N$ and $L$ together so that $N/L^2$ remains fixed tests size more cleanly in the Vicsek model. In a real flock, however, density, shape and interaction structure may change with group size, and that coupling may itself be scientifically important.
